```python
import random
```

```python
harcamalar = []
```

```python
harcama = {    "isim": "Kahve",    "kategori": "Yeme İçme",    "tutar": 120.0}
```

```python
harcamalar.append(harcama)
```

```python
#input("Aylik butcenizi giriniz: ")
```

```python
butce = float(input("Aylik butcenizi giriniz: ")) 
```

```python
butce
```

```python
while True:
    print("1 - Harcama Ekle")
    print("2 - Harcamaları Listele")
    print("3 - Harcama Analizi")
    print("4 - Kategori Ara")
    print("5 - Rastgele Tasarruf Tavsiyesi")
    print("6 - Cikis\n\n")

    secim = input("Yapmak istediğiniz işlemi seçiniz (1-6): ")

    if secim == "1":
        harcama = {"isim": input("Harcama adi:"), "kategori": input("Kategori:"), "tutar": float(input("Tutar:"))}
        harcamalar.append(harcama)

    if secim == "2":
        for sira, harcama in enumerate(harcamalar, 1):
            print(f"{sira}. {harcama['isim']} | {harcama['kategori']} | {harcama['tutar']} TL")

    if secim == "3":
        toplam = 0
        tutarlar = []
        for harcama in harcamalar:
            toplam += harcama["tutar"]
            tutarlar.append(harcama["tutar"])
        print(f"Toplam Harcama: {toplam}")
        print(f"Butce: {butce}")
        print(f"Kalan Butce:{butce - toplam}\n\n ")

        en_yuksek = harcamalar[tutarlar.index(max(tutarlar))]
        en_dusuk = harcamalar[tutarlar.index(min(tutarlar))]
        print(f"En Yuksek Harcama:\n {en_yuksek['isim']}-{en_yuksek['tutar']}")
        print(f"En Dusuk Harcama: \n {en_dusuk['isim']}-{en_dusuk['tutar']}") #yz

    if secim == "4":
        oran = butce / toplam
        if oran<50 :
            print("Butcen gayet iyi durumda.")
        if oran>50 and oran<80:
            print("Harcamalarına dikkat et.")
        if oran>80 and oran<100:
            print("Butce sinirina yaklastin.")
        elif oran>100:
            print("Butceni astin!!!")

    if secim == "5":
        tavsiyeler = [    "Bu hafta dışarıdan kahve almak yerine evden götür.",    "Kullanmadığın abonelikleri kontrol et.",    "Market alışverişine listeyle çık.",    "Haftalık harcama limiti belirle."]
        rastgele_secim = random.choice(tavsiyeler)
        print(f"Bugunun tavsiyesi:\n {rastgele_secim}")
        
    elif secim == "6":
        break

    else:
        print("Geçersiz seçim! Lütfen 1-6 arasında bir sayı seciniz")
```

```python
tupleListe = []
for harcama in harcamalar:
    tupleListe.append((harcama["tutar"], harcama["isim"], harcama["kategori"]))

siraliHarcamalar = sorted(tupleListe, reverse=True)
print("En pahali uc harcama:")
for tutar, isim, kategori in siraliHarcamalar[:3]:
    print(f"{isim} | {kategori} | {tutar} TL") #yz
```
